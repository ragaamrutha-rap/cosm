```python
import whisper
from moviepy.editor import VideoFileClip
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
video_path = "videoplayback.mp4"

video = VideoFileClip(video_path)
audio_path = "audio.wav"

video.audio.write_audiofile(audio_path)
```

```python
model = whisper.load_model("base")

result = model.transcribe("audio.wav")

transcribed_text = result["text"]

print(transcribed_text)
```

```python
reference_text = """
A one-minute introduction can make people remember you forever if done the right way. Instead of simply stating your name, job title, or where you are from, you can introduce yourself in a way that creates a real connection. The key is not speaking more, but speaking meaningfully. A simple three-part structure can help: the hook, the proof, and the invitation. The hook replaces your job title with what you actually do and who you help, making it more interesting and relatable. For example, instead of saying “I’m a financial analyst,” you could say you help young families protect their biggest investment, their home. The proof briefly explains why you are passionate about what you do or what motivates you, adding credibility and personality. Finally, the invitation turns your introduction into a conversation by asking an open-ended question, such as asking the other person what they are passionate about or what they like to build or solve. This approach helps create genuine engagement and makes your introduction memorable.
"""
```

```python
similarity_model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
reference_embedding = similarity_model.encode([reference_text])
transcribed_embedding = similarity_model.encode([transcribed_text])
```

```python
similarity = cosine_similarity(reference_embedding, transcribed_embedding)

score = similarity[0][0] * 100

print("Similarity Score:", score, "%")
```

```python
import cv2
import os

video_path = "videoplayback.mp4"
frame_dir = "frames"
os.makedirs(frame_dir, exist_ok=True)

video = cv2.VideoCapture(video_path)
frame_rate = 1  # extract 1 frame per second
count = 0
frame_count = 0

fps = int(video.get(cv2.CAP_PROP_FPS))

while video.isOpened():
    ret, frame = video.read()
    if not ret:
        break
    if count % fps == 0:  # 1 frame per second
        frame_file = os.path.join(frame_dir, f"frame_{frame_count}.jpg")
        cv2.imwrite(frame_file, frame)
        frame_count += 1
    count += 1

video.release()
print(f"Extracted {frame_count} frames.")
```

```python
!pip install ultralytics

from ultralytics import YOLO
import glob

# Load pre-trained YOLOv8 model
model = YOLO("yolov8n.pt")  # nano model for speed

frames = glob.glob("frames/*.jpg")
semantic_data = []

for frame_file in frames:
    results = model(frame_file)
    detected_objects = [box.cls for box in results[0].boxes]  # class IDs
    names = [model.names[int(cls)] for cls in detected_objects]
    semantic_data.append({
        "frame": frame_file,
        "objects": names
    })

print(semantic_data[:3])  # show first 3 frames
```

```python
actions = []

for data in semantic_data:
    objs = data["objects"]
    action_list = []
    if "person" in objs and "door" in objs:
        action_list.append("person enters room")
    if "cat" in objs and "table" in objs:
        action_list.append("cat jumps on table")
    actions.append({
        "frame": data["frame"],
        "actions": action_list
    })
```

```python
semantic_summary = []

for frame_objects, frame_actions in zip(semantic_data, actions):
    text = "Frame {}: Objects detected - {}. Actions - {}".format(
        os.path.basename(frame_objects["frame"]),
        ", ".join(frame_objects["objects"]) if frame_objects["objects"] else "None",
        ", ".join(frame_actions["actions"]) if frame_actions["actions"] else "No significant action"
    )
    semantic_summary.append(text)

# Print semantic summary
for line in semantic_summary:
    print(line)
```
